# Original idea revisited: feature-sequence LSTM

This retains the original idea of feeding feature columns as timesteps. It is a methodological comparison, not a forecast through transaction history. Its arbitrary feature ordering is a limitation.

```python
import os
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
SMOKE = os.environ.get("HDB_SMOKE") == "1"
DATA = ROOT / ("tests/fixtures/transactions.csv" if SMOKE else "data/raw/resale.csv")
RUNS = Path(os.environ.get("HDB_RUN_ROOT", str(ROOT / "artifacts")))
```

## Optional experiment

Install `requirements-neural.txt` after the base requirements. Core baselines do not depend on PyTorch. We use bounded CPU runs for this portfolio; CUDA is optional. Target scaling is fitted only on training prices. Checkpoints use validation MAE, not training loss.

```python
from hdb_resale.experiment import (
    evaluate_experiment,
    load_bundle,
    predict,
    train_experiment,
)

run = RUNS / "lstm"
if not (run / "metadata.json").exists():
    train_experiment(
        DATA, "lstm", run, epochs=2 if SMOKE else 5, threads=2, device="cpu"
    )
evaluate_experiment(run, DATA)
```

```python
import pandas as pd

history = pd.read_csv(run / "history.csv")
history
```

## Checkpoint selection

The saved weights belong to the epoch with lowest validation MAE. The standardised target is converted back to SGD using training-only statistics. CPU inference reconstructs the network from a state dictionary rather than unpickling a notebook-defined model class.

```python
from hdb_resale.data import load_data

sample = load_data(DATA).tail(3).drop(columns="resale_price")
sample.assign(predicted_price_sgd=predict(load_bundle(run), sample))
```

## Interpretation limits

Four LSTM layers with 64 hidden units process 10 feature positions. Categorical values use training-fitted ordinal encoding and scaling to preserve the original compact feature-sequence idea; those codes do not imply a meaningful category order. The corrected regression head is linear and the target is standardised.

This five-epoch run is deliberately budget-limited. It is not evidence about the best possible LSTM, and it is not a temporal LSTM. Fair data splits do not make unequal training budgets equivalent.
